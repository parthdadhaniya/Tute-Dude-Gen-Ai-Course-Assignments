## Setup: Model & LangChain Imports

```python
from config import get_chat_model
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

model = get_chat_model()
res = model.invoke([HumanMessage(content="Hi")])
print("Model Response:", res.content)
```

## Part 1: Messages & MessagesPlaceholder

### Task 1: Conceptual Answers
- **SystemMessage**: Defines the overall behavior, persona, and rules for the chatbot.
- **HumanMessage**: The message coming from the user.
- **AIMessage**: The reply generated by the AI model.
- **Why message-based prompting is better**: It separates roles clearly, avoids prompt injection, and makes it easy to append conversational history.

### Task 2: MessagesPlaceholder Usage

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a friendly customer support assistant."),
    MessagesPlaceholder(variable_name="chat_history"),
    ("human", "{user_input}")
])

chain = prompt | model
chat_history = []

turns = [
    "Hi, my name is Parth.",
    "What is my name?",
    "Explain Python lists in simple words."
]

for turn_num, user_msg in enumerate(turns, 1):
    print(f"\nTurn {turn_num}: User -> {user_msg}")
    response = chain.invoke({
        "chat_history": chat_history,
        "user_input": user_msg
    })
    print(f"AI: {response.content}")
    chat_history.append(HumanMessage(content=user_msg))
    chat_history.append(AIMessage(content=response.content))

print(f"\nTotal history count: {len(chat_history)} messages")
```
